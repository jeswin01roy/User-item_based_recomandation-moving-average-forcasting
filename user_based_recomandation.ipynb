```python
import pandas as pd 
from sklearn.metrics.pairwise import cosine_similarity
```

```python
ratings = pd.DataFrame({

    "Avengers" : [5,5,4,4,0],
    "Iron Man" : [4,5,5,4,0],
    "Titanic"  : [1,1,5,2,5],
    "Avatar"   : [2,2,5,3,4],
    "Frozen"   : [0,5,4,0,5],
    "John Wick": [5,4,0,5,5]

},

index=["User1","User2","User3","User4","User5"])


print("User-Item Rating Matrix\n")

print(ratings)
```

```python
user_similarity = cosine_similarity(ratings)
```

```python
similarity_df = pd.DataFrame(

    user_similarity,

    index=ratings.index,

    columns=ratings.index

)

print("\nUser Similarity Matrix\n")

print(similarity_df.round(3))
```

```python
def recommend_movies(target_user):

    print("\n================================================")

    print("Recommendations for",target_user)

    print("================================================")

    similar_users = similarity_df[target_user].sort_values(
        ascending=False
    )

    print("\nSimilarity Scores\n")

    print(similar_users.round(3))


    similar_users = similar_users.drop(target_user)


    nearest_user = similar_users.index[0]

    print("\nNearest User :",nearest_user)


    target_ratings = ratings.loc[target_user]

    nearest_ratings = ratings.loc[nearest_user]


    recommendations=[]

    for other_user in similar_users.index:
        
        other_ratings =ratings.loc[other_user]

        for movie in ratings.columns:
        
            if (target_ratings[movie]==0 and
                other_ratings[movie]>=4 and 
                movie not in recommendations):
            
                recommendations.append(movie)



    if len(recommendations)==0:

        print("\nNo Recommendations Found.")
            
        

    else:

        print("\nRecommended Movies\n")

        for movie in recommendations:

            print(movie)
```

```python
recommend_movies("User1")

recommend_movies("User3")

recommend_movies("User5")
```
