```python
# ==========================================================
# ITEM-BASED COLLABORATIVE FILTERING
# ==========================================================
# In Item-Based Collaborative Filtering,
# recommendations are made based on the similarity
# between items (movies).
#
# Example:
# If a user likes "Avengers",
# recommend movies that are similar to "Avengers".
#
# Similarity is calculated using the ratings
# given by all users.
# ==========================================================


# ==========================================================
# STEP 1 : Import Required Libraries
# ==========================================================

import pandas as pd

# Used to calculate cosine similarity
from sklearn.metrics.pairwise import cosine_similarity

```

```python

# ==========================================================
# STEP 2 : Create User-Item Rating Matrix
# ==========================================================
# Rows    -> Users
# Columns -> Movies
# Values  -> Ratings (1-5)
#
# 0 means the user has not rated the movie.

ratings = pd.DataFrame({

    "Avengers":[5,5,0,4,4],
    "Iron Man":[4,5,4,5,0],
    "Titanic":[1,0,5,2,5],
    "Avatar":[2,1,5,2,4],
    "Frozen":[0,4,4,0,5],
    "John Wick":[5,4,0,5,5]

},

index=["User1","User2","User3","User4","User5"])


print("User-Item Rating Matrix")
print(ratings)



```

```python
# ==========================================================
# STEP 3 : Transpose the Matrix
# ==========================================================
# For Item-Based Filtering,
# movies should become rows.

movie_matrix = ratings.T

print("\nMovie Matrix")
print(movie_matrix)



```

```python
# ==========================================================
# STEP 4 : Calculate Movie Similarity
# ==========================================================
# Cosine similarity is calculated
# between every pair of movies.

movie_similarity = cosine_similarity(movie_matrix)

# Convert similarity matrix into DataFrame
similarity_df = pd.DataFrame(

    movie_similarity,

    index=movie_matrix.index,

    columns=movie_matrix.index

)

print("\nMovie Similarity Matrix")
print(similarity_df)


```

```python
# ==========================================================
# STEP 5 : Create Recommendation Function
# ==========================================================

# Function to recommend movies similar to the given movie
# Input : movie_name (movie entered by the user)
# Output: Top 5 similar movies with similarity scores
def recommend_movies(movie_name):

    # Print the movie for which recommendations are generated
    print("\nRecommendations for", movie_name)

    # Print a separator line for better readability
    print("------------------------------------------")

    # ---------------------------------------
    # Get similarity scores of the movie
    # ---------------------------------------

    # Select the column corresponding to the entered movie
    # from the similarity matrix (DataFrame).
    #
    # Example:
    # movie_name = "Avatar"
    #
    # Avatar         1.000
    # Titanic        0.842
    # Avengers       0.721
    # Frozen         0.356
    #
    # This returns a Pandas Series where:
    # Index  = Movie names
    # Values = Similarity scores
    similar_movies = similarity_df[movie_name]

    # ---------------------------------------
    # Sort similarity scores
    # Highest similarity first
    # ---------------------------------------

    # Sort similarity scores in descending order.
    #
    # ascending=False means:
    # Highest similarity score appears first.
    #
    # Before sorting:
    # Frozen      0.356
    # Avatar      1.000
    # Titanic     0.842
    #
    # After sorting:
    # Avatar      1.000
    # Titanic     0.842
    # Frozen      0.356
    similar_movies = similar_movies.sort_values(
        ascending=False
    )

    # Print all similarity scores
    print("\nSimilarity Scores")

    # Display the sorted similarity scores
    print(similar_movies)

    # ---------------------------------------
    # Remove the movie itself
    # Similarity with itself = 1
    # ---------------------------------------

    # Every movie is perfectly similar to itself.
    # Therefore, its similarity score is always 1.
    #
    # We don't recommend the same movie,
    # so remove it from the list.
    similar_movies = similar_movies.drop(movie_name)

    # ---------------------------------------
    # Display Top 5 Similar Movies
    # ---------------------------------------

    # Print heading
    print("\nTop Recommended Movies")

    # head(5) returns only the first five movies
    # (highest similarity scores after removing
    # the selected movie).
    #
    # items() returns each row as:
    # (movie_name, similarity_score)
    #
    # Example:
    # ("Titanic", 0.842)
    # ("Avengers", 0.721)
    for movie, score in similar_movies.head(5).items():

        # Print movie name and similarity score
        # round(score, 3) displays the score
        # with three decimal places.
        print(movie, "Similarity :", round(score, 3))
```

```python
# ==========================================================
# STEP 6 : Test the Recommendation System
# ==========================================================

recommend_movies("Avengers")

recommend_movies("Titanic")

recommend_movies("Frozen")
```
